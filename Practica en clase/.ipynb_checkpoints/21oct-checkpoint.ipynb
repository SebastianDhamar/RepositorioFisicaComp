```python
import numpy as np
import matplotlib.pyplot as plt

def H(n,x):
    if n==0:
        return 1
    elif n==1:
        return 2*x
    elif n>1:
        return (2*x*H(n-1,x))-(2*(n-1)*H(n-2,x))
```

```python
H(3,4)
```

```python
import math as mt
def psi(n,x):
    if n>=0:
        return (1/np.sqrt((2**n)*mt.factorial(n)*np.sqrt(np.pi)))*np.exp(-(x**2)/2)*H(n,x)
```

```python
N = [0,1,2,3]
X= np.linspace(-4,4,100)

Valores = [[],[],[],[]]
for n in N:
    for x in X:
        v=psi(n,x)
        Valores[n].append(v)

plt.plot(X,Valores[0],"r",label="$H_0(x)$")
plt.plot(X,Valores[1],"b", label="$H_1(x)$")
plt.plot(X,Valores[2],"g",label="$H_2(x)$")
plt.plot(X,Valores[3],"y",label="$H_3(x)$")
plt.legend()
plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')
plt.grid()
plt.show()
```
