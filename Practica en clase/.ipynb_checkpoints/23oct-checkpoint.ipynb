## Diferenciacuion Numerica

Diferencias centrales

```python
n=9
h=10**(-n)

def f(x):
    return np.sin(x)

def f2(x,y):
    return x**2 + y**2

def df(x):
    return (f(x+h/2)-f(x-h/2))/h

def d2f(x):
    return(f(x+h)-2*f(x)-f(x-h))/(h**2)

def dpx(x,y):
    return(f2(x+h/2,y)-f2(x-h/2,y))/h

def dpy(x,y):
    return(f2(x,y+h/2)-f2(x,y-h/2))/h
```

```python
dpx(3,5)
```

```python
dpy(5,3)
```

```python
import numpy as np
import matplotlib.pyplot as plt

X=np.linspace(-np.pi,np.pi,100)
Y=f(X)
Yp=df(X)

plt.figure(figsize=(9,6))

plt.subplot(2,1,1)
plt.plot(X,Y,"r")
plt.title('Original',fontsize=13)
plt.grid()
plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')

plt.subplot(2,1,2)
plt.plot(X,Yp,"b")
plt.title("Derivada",fontsize=13)
plt.grid()
plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')


plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X=np.linspace(-np.pi,np.pi,100)
Y=f(X)
Yp=d2f(X)

plt.figure(figsize=(9,6))

plt.subplot(2,1,1)
plt.plot(X,Y,"r")
plt.title('Original',fontsize=13)
plt.grid()
plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')

plt.subplot(2,1,2)
plt.plot(X,Yp,"b")
plt.title("Derivada",fontsize=13)
plt.grid()
plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')


plt.show()
```

Diferencia hacia adelante

```python
n=9
h=10**(-n)

def f(x):
    return np.sin(x)

def df(x):
    return (f(x+h)-f(x))/h
```

```python
df(0)
```

Diferencia hacia atras

```python
n=9
h=10**(-n)

def f(x):
    return np.sin(x)

def df(x):
    return (f(x)-f(x-h))/h
```

```python
df(0)
```
