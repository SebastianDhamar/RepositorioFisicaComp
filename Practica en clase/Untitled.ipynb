```python
Datos = open("data/Ising 100.csv", "r")

Valores = []
for n in Datos:
    Valores.append(n.split())

Temp=[]
Energia=[]
Magne=[]
CapCal=[]
SusMag=[]

for n in range(len(Valores)):
    Temp.append(Valores[n][0])
for n in range(len(Valores)):
    Energia.append(Valores[n][1])
for n in range(len(Valores)):
    Magne.append(Valores[n][2])
for n in range(len(Valores)):
    CapCal.append(Valores[n][3])
for n in range(len(Valores)):
    SusMag.append(Valores[n][4])

print("Guardados")
```

```python
import numpy as np
Dom=[0,45,90,135,180,225,270,315,360]
Sin=[]

for i in Dom:
    Sen= float(np.sin(np.radians(i)))
    Sen = round(Sen,5)
    Sin.append(Sen)

# Sin
escritura = open("data/DomSin", "w")
for n in range(len(Dom)):
    Linea = str(Dom[n]),str(Sin[n])
    Linea = str(Linea)
    escritura.writelines(Linea)

escritura.close()
```

```python
def factorr(n):
    if n == 1:
        return 1
    else:
        return n*factorr(n-1)
```

```python
factorr(2)
```

```python
import numpy as np
def catr(n):
    if n ==1:
        return 1
    else:
        return (((4*n)-2)/(n+1))*catr(n-1)
```
