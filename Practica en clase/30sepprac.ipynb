1- Graficar en el plano xy la curva en coordenadas polares correspondiente a la funcion de Fey.

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 24*np.pi, 10000)

r = (np.e**(np.cos(theta)))-(2*np.cos(4*theta))+((np.sin(theta/12))**5)

x = r*(np.cos(theta))
y = r*(np.sin(theta))

plt.plot(x, y, color='purple')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Funcion de Fey")
plt.axhline(y = 0.0, xmin = 0.0, color = 'black')
plt.axvline(x = 0.0, ymin = 0.0, color = 'black')
plt.grid()

plt.show()
```

2- En una grafica de calor mostrar el potencial electrostatico de una particula con carga de 1 coulomb en el centro de un plano de un metro cuadrado.

```python
import numpy as np
import matplotlib.pyplot as plt
k=8.99*(10**9)
x = np.linspace(-0.5, 0.5, 100)
y = np.linspace(-0.5, 0.5, 100)
Q = 1
X,Y=np.meshgrid(x,y)

r = np.sqrt((X**2)+(Y**2))

V= k*(Q/r)
plt.imshow(V)
```
