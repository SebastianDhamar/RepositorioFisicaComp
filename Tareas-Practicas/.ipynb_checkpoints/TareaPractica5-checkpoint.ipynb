Crea una función `f(x)` que devuelva $1 + \frac{1}{2} \tanh 2x$, luego usa **diferencias centrales** para calcular numéricamente la derivada de la función en el intervalo $-2 \le x \le 2$.

$$ \frac{d}{dx}(1+\frac{1}{2}tanh(2x)) = 0+ \frac{1}{2}\frac{d}{dx}(tanh(2x))$$
$$ = \frac{1}{2}(2sech^2(2x)) = sech^2(2x)$$

```python
import math as mt
import numpy as np

def f(x):
    return 1+ (1/2)*mt.tanh(2*x)

def df(x):
    return (f(x+h/2)-f(x-h/2))/h

def sech(x):
    return 1/np.cosh(x)
```

2. **Campo eléctrico de una distribución de cargas** (valor 3 puntos).

Supongamos que tenemos una distribución de cargas y queremos calcular el campo eléctrico resultante. Una forma de hacerlo es calcular primero el potencial eléctrico $\phi$ y luego tomar su gradiente.

Para una carga puntual $q$ en el origen, el potencial eléctrico a una distancia $r$ del origen es $\phi = q / 4\pi\epsilon_0 r$ y el campo eléctrico es $\vec{E} = -\nabla\phi$.

(a) Suponiendo que tienes dos cargas, de $+1\text{ C}$ y $-1\text{ C}$ (respectivamente), separadas $10\text{ cm}$. Calcula el potencial eléctrico resultante en un plano cuadrado de $1\text{ m} \times 1\text{ m}$ que rodea las cargas y pasa a través de ellas. Calcula el potencial en puntos espaciados a $1\text{ cm}$ en una cuadrícula y haz una visualización en la pantalla del potencial usando un gráfico de densidad.

<div align="center">
    <img src="img/t5e2a.png">
</div>

```python
import numpy as np
def phi(q,r):
    e0= 8.85*10**(-12)
    return q/(4*np.pi*e0*r)
```

```python
import matplotlib.pyplot as plt

d=0.1
q1= 1
x1, y1 = -d / 2, 0.0
q2=-1
x2, y2 =  d / 2, 0.0

x=np.linspace(-0.5,0.5,300)
y=np.linspace(-0.5,0.5,300)
X,Y=np.meshgrid(x,y)

r1= np.sqrt((X-x1)**2 +(Y-y1)**2)
r2= np.sqrt((X-x2)**2 +(Y-y2)**2)

P1= phi(q1,r1)
P2= phi(q2,r2)

P= P1+P2

plt.imshow(P, extent=[-0.5, 0.5, -0.5, 0.5], origin="lower", cmap="RdBu_r", vmin=-1e10, vmax=1e10)
plt.colorbar(label="Potencial $\phi$ (V)")
plt.xlabel("x(m)")
plt.ylabel("y(m)")
plt.show()
```

Calcula y visualiza el campo eléctrico resultante en puntos espaciados a $1\text{ cm}$ en 1 metro cuadrado del plano $xy$ para el caso donde $L = 10\text{ cm}$. La distribución de carga está centrada en el medio del área visualizada y $q_0 = 100\,\frac{\text{C}}{\text{m}^2}$.

Tendrás que realizar una integral doble sobre $x$ y $y$, luego diferenciar el potencial respecto a la posición para obtener el campo eléctrico. Elige cualquier método de integración que parezca apropiado para las integrales.

```python
L = 0.1
N = 30
a= -L/2
b = L/2
e0= 8.85*10**(-12)

xi, wi = np.polynomial.legendre.leggauss(N)
xp = 0.5 * L * xi
yp = 0.5 * L * xi
wp = 0.5 * L * wi
Xp, Yp = np.meshgrid(xp, yp)
Wp = np.outer(wp, wp)
q0=100
L=0.1
sigmap = q0 * np.sin(2*np.pi*Xp/L)*np.sin(2*np.pi*Yp/L)
dq = sigmap * Wp

def phiTot(x, y):
    potencial = np.zeros_like(x)
    for i in range(N):
        for j in range(N):
            r= np.sqrt((x-xp[i])**2+(y-yp[j])**2)
            r= np.where(r == 0,1e-4,r)
            potencial += dq[j,i]/r
    return (1/(4*np.pi*e0))*potencial

h=10**(-5)
def dpx(x,y):
    return(phiTot(x+h/2,y)-phiTot(x-h/2,y))/h
def dpy(x,y):
    return(phiTot(x,y+h/2)-phiTot(x,y-h/2))/h

x = np.linspace(-0.5, 0.5, 101)
y = np.linspace(-0.5, 0.5, 101)
X, Y = np.meshgrid(x, y)

Ex= -dpx(X,Y)
Ey= -dpy(X,Y)
E_mag = np.sqrt(Ex**2 + Ey**2)
E_dir = np.arctan2(Ey,Ex)


fig, axs = plt.subplots(1, 2, figsize=(13, 5))

im0 = axs[0].imshow(E_mag, extent=[-0.5, 0.5, -0.5, 0.5], cmap='inferno', vmax=1e11)
axs[0].set_title(r'Magnitud |E|')
axs[0].set_xlabel('x (m)')
axs[0].set_ylabel('y (m)')
fig.colorbar(im0, ax=axs[0], label=r'$|\vec{E}|$ (V/m)')

im1 = axs[1].imshow(E_dir, extent=[-0.5, 0.5, -0.5, 0.5], cmap='hsv', vmin=-np.pi, vmax=np.pi)
axs[1].set_title('Dirección del campo eléctrico')
axs[1].set_xlabel('x(m)')
axs[1].set_ylabel('y(m)')
cbar1 = fig.colorbar(im1, ax=axs[1], label='Ángulo (rad)')
cbar1.set_ticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
cbar1.set_ticklabels([r'$-\pi$', r'$-\pi/2$', r'$0$', r'$\pi/2$', r'$\pi$'])
plt.show()
```

### 3. **Procesamiento de imágenes y *STM*** (valor 4 puntos)

Cuando la luz incide sobre una superficie, la cantidad que cae por unidad de área depende no sólo de la intensidad de la luz, sino también del ángulo de incidencia. Si la luz forma un ángulo $\theta$ con la normal, sólo "ve" una fracción $\cos\theta$ de área, por unidad de área real en la superficie:

<div align="center">
  <img src="img/t5e3a.png" width="350"/>
</div>

Así, la intensidad de la iluminación es $a\cos\theta$, si $a$ es la intensidad bruta de la luz. Esta simple ley física es un elemento central de los gráficos por computadora en 3D. Nos permite calcular cómo incide la luz sobre objetos tridimensionales y, por tanto, cómo se verán cuando se iluminen desde varios ángulos.

Supongamos, por ejemplo, que miramos la Tierra desde arriba y vemos sus montañas. Conocemos la altura de las montañas $w(x, y)$ en función de la posición en el plano, por lo que la ecuación para la superficie de la Tierra es simplemente $z = w(x, y)$, o equivalentemente $w(x, y) - z = 0$, y el vector normal $\vec{v}$ a la superficie está dado por el gradiente de $w(x, y) - z$ de la siguiente manera:

$$
\vec{v} = \nabla [w(x, y) - z] = \begin{pmatrix} \partial / \partial x \\ \partial / \partial y \\ \partial / \partial z \end{pmatrix} [w(x, y) - z] = \begin{pmatrix} \partial w / \partial x \\ \partial w / \partial y \\ -1 \end{pmatrix}
$$

Ahora supongamos que tenemos luz entrante representada por un vector $\vec{a}$ con magnitud igual a la intensidad de la luz. Entonces el producto escalar de los vectores $\vec{a}$ y $\vec{v}$ es:

$$
\vec{a} \cdot \vec{v} = |\vec{a}| \, |\vec{v}| \cos\theta,
$$

donde $\theta$ es el ángulo entre los vectores.

Entonces, la intensidad de la iluminación de la superficie de las montañas es:

$$
I = |\vec{a}| \cos\theta = \frac{\vec{a} \cdot \vec{v}}{|\vec{v}|} = \frac{a_x (\partial w / \partial x) + a_y (\partial w / \partial y) - a_z}{\sqrt{(\partial w / \partial x)^2 + (\partial w / \partial y)^2 + 1}}
$$

Tomemos un caso simple donde la luz brilla horizontalmente con intensidad unitaria, a lo largo de una línea en un ángulo $\phi$ en sentido contrario a las manecillas del reloj desde el eje este-oeste, de modo que $\vec{a} = (\cos\phi, \sin\phi, 0)$. Entonces nuestra intensidad de iluminación se simplifica a:

$$
I = \frac{\cos\phi \, (\partial w / \partial x) + \sin\phi \, (\partial w / \partial y)}{\sqrt{(\partial w / \partial x)^2 + (\partial w / \partial y)^2 + 1}}
$$

Así, si podemos calcular las derivadas de la altura $w(x, y)$ y sabemos $\phi$, entonces podemos calcular la intensidad en cualquier punto.

Para calcular las derivadas, necesitarás saber el valor de $h$ y la distancia en metros entre puntos de la cuadrícula, que es de aproximadamente $30{,}000\text{ m}$ en este caso$^1$.

```python
import numpy as np
w = np.loadtxt("data/altitudes.txt")
h=30000
Ny, Nx = w.shape

def dwx(w, h):
    dwpx= np.zeros((Ny,Nx))
    dwpx[:,0]= (w[:,1]-w[:,0])/h
    dwpx[:,1:-1]= (w[:,2:]-w[:,:-2])/(2*h)
    dwpx[:,-1]=(w[:,-1]-w[:,-2])/h
    return dwpx

def dwy(w,h):
    dwpy= np.zeros((Ny,Nx))
    dwpy[0,:]= (w[1,:]-w[0,:])/h
    dwpy[1:-1,:]= (w[2:,:]-w[:-2,:])/(2*h)
    dwpy[-1,:]=(w[-1,:]-w[-2,:])/h
    return dwpy

dw_dx = dwx(w, h)
dw_dy = dwy(w, h)

print(dw_dx, "dimensiones=",dw_dx.shape)
print(dw_dy, "dimensiones=",dw_dy.shape)
```

Para las fronteras iniciales, es decir primer fila y primer columna de la matriz, use el metodo de la diferencia hacia adelante, pues no hay un valor previo para poder aplicar centrales o hacia atras, para la frontera final, ultima fila y ultima columna, use el metodo de diferencia hacia atras por el mismo motivo, no hay valores posteriores para usar diferencias centrales o hacia adelante, finalmente para los valores interiores de la matriz use diferencias centrales pues suele ser mas preciso, pero como no hay valores medios, lo aplique dando saltos de 2h.

### 4. Derivación mediante integración. (valor 2 puntos).

Supongamos que tenemos una función $f(z)$ cuyo valor conocemos no solo sobre la recta real, sino también para valores complejos de su argumento. Entonces, podemos calcular las derivadas de dicha función en cualquier punto $z_0$ mediante una integral de contorno, utilizando la fórmula de la derivada de Cauchy:

$$
\left( \frac{\mathrm{d}^m f}{\mathrm{d}z^m} \right)_{z=z_0} = \frac{m!}{2\pi i} \oint \frac{f(z)}{(z - z_0)^{m+1}} \, \mathrm{d}z,
$$

donde la integral se realiza en sentido antihorario a lo largo de cualquier contorno en el plano complejo que rodee al punto $z_0$, pero que no contiene polos de $f(z)$.

Dado que la integración numérica es considerablemente más sencilla y precisa que la diferenciación numérica, esta fórmula nos proporciona un método para calcular derivadas con gran precisión (especialmente de orden superior), transformándolas en integrales.

Por ejemplo, supongamos que queremos calcular las derivadas de $f(z)$ en $z = 0$. Aplicando la fórmula de Cauchy y utilizando la regla del trapecio, para calcular la integral a lo largo de un contorno circular centrado en el origen con radio 1. La regla del trapecio será ligeramente distinta a la versión a la que hemos trabajado, ya que el valor del intervalo $h$ es ahora un número complejo; además, este no permanece constante entre los distintos segmentos de la integral: su módulo, sí se mantiene constante, pero su argumento varía de un segmento a otro.

Dividiendo la integral en $N$ segmentos, con puntos muestra $z_k$ distribuidos uniformemente a lo largo del contorno circular, en las posiciones $z_k = \mathrm{e}^{i 2\pi k / N}$, para $k = 0, \dots, N$. Entonces, la distancia entre puntos de muestreo consecutivos es:

$$
h_k = z_{k+1} - z_k = \mathrm{e}^{i 2\pi (k+1) / N} - \mathrm{e}^{i 2\pi k / N},
$$

y, definiendo $g(z) = f(z) / z^{m+1}$ con notación abreviada para el integrando, la aproximación de la integral mediante la regla del trapecio sería:

$$
\begin{aligned}
\oint g(z) \, \mathrm{d}z &\simeq \sum_{k=0}^{N-1} \frac{1}{2} \left[ g(z_{k+1}) + g(z_k) \right] \left[ \mathrm{e}^{i 2\pi (k+1) / N} - \mathrm{e}^{i 2\pi k / N} \right] \\
&= \frac{1}{2} \left[ \sum_{k=0}^{N-1} g(z_{k+1})\,\mathrm{e}^{i 2\pi (k+1) / N} - \sum_{k=0}^{N-1} g(z_k)\,\mathrm{e}^{i 2\pi k / N} \right. \\
&\quad\quad \left. - \sum_{k=0}^{N-1} g(z_{k+1})\,\mathrm{e}^{i 2\pi k / N} + \sum_{k=0}^{N-1} g(z_k)\,\mathrm{e}^{i 2\pi (k+1) / N} \right].
\end{aligned}
$$

Si observamos que $z_N = z_0$, las dos primeras sumas dentro de los corchetes se cancelan por completo, y las dos sumas restantes son iguales, salvo por factores de fase triviales; por tanto, toda la expresión se simplifica a:

$$
\begin{aligned}
\oint g(z) \, \mathrm{d}z &\simeq \frac{1}{2} \left[ \mathrm{e}^{i 2\pi / N} - \mathrm{e}^{-i 2\pi / N} \right] \sum_{k=0}^{N-1} g(z_k)\,\mathrm{e}^{i 2\pi k / N} \\
&\simeq \frac{2\pi i}{N} \sum_{k=0}^{N-1} f(z_k)\,\mathrm{e}^{-i 2\pi k m / N},
\end{aligned}
$$

donde hemos vuelto a utilizar la definición de $g(z)$. Combinando este resultado con la fórmula de Cauchy, obtenemos que:

$$
\left( \frac{\mathrm{d}^m f}{\mathrm{d}z^m} \right)_{z=0} \simeq \frac{m!}{N} \sum_{k=0}^{N-1} f(z_k)\,\mathrm{e}^{-i 2\pi k m / N}. \tag{1}
$$

Es fácil demostrar que el valor correcto de la $m$-ésima derivada en este caso es $2^m$; por tanto, resulta sencillo comprobar si tu programa funciona correctamente: los resultados deberían ser potencias de dos ($2, 4, 8, 16, 32$, etc.). Notarás que, mediante esta técnica, es posible obtener resultados razonablemente precisos para las veinte derivadas de forma rápida. Por otro lado, si utilizas las fórmulas de *diferencias* estándar para calcular las derivadas, observarás que solo es posible determinar con precisión las tres o cuatro primeras antes de que los errores numéricos se vuelvan tan grandes que los resultados carezcan de utilidad. En este caso, la *fórmula de Cauchy* proporciona, por consiguiente, mejores resultados.

Por cierto, la suma $\sum_k f(z_k)\,\mathrm{e}^{i 2\pi k m / N}$ que aparece en la fórmula anterior se denomina *transformada discreta de Fourier* de los puntos muestra complejos $f(z_k)$. Existe una técnica elegante para evaluar la *transformada de Fourier* para muchos valores de $m$ de forma simultánea, conocida como *transformada rápida de Fourier*, la cual puede resultar útil en situaciones en las que la evaluación directa de la fórmula es lenta.

```python
import math as mt
import cmath as cmt

N=10000
M=np.arange(1,21,1)

def z(k,N):
    return cmt.exp(1j*2*np.pi*k/N)

def fm(z):
    return cmt.exp(2*z)

def dmf0(f,m,z,N):
    S=0+0j
    for k in range(N):
        S+=f(z(k,N))*cmt.exp(-1j*2*np.pi*k*m/N)
    return (mt.factorial(m)/N)*S

for i in M:
    print("derivada ",i,"=",dmf0(fm,i,z,N))
```
