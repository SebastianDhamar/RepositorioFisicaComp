```python
import numpy as np

a1=15.8
a2=18.3
a3=0.714
a4=23.2

A=float(input("Valor de A: "))
Z=float(input("Valor de Z: "))

if A%2!=0:
    a5=0.0
elif A%2==0 and Z%2==0:
    a5=12.0
elif A%2==0 and Z%2!=0:
    a5=-12.0

B=(a1*A)-(a2*(A**(2/3)))-(a3*((Z**2)/(A**(1/3))))-(a4*(((A-(2*Z))**2)/A))+a5/(A**(1/2))

print(B,"MeV")
```

3. Coeficientes binomiales
El coeficiente binomial ($\frac{n}{k}$) es un numero entero igual a:

<div align="center">
    <img src="img/ec3t2.png">
</div>

donde k ≥ 1, o bien (($\frac{n}{k}$) = 1 cuando k = 0.


a) Utiliza esta formula para escribir una función llamada binomial(n,k) (o como tu quieras) que calcule el coeficiente binomial para un n y k dados. Asegurate de que tu función devuelva la respuesta en forma de un número entero (no flotante) y proporcione el valor correcto de 1 para el caso en que k = 0.

```python
def factorr(n):
    if n == 1:
        return 1
    else:
        return n*factorr(n-1)
```

```python
def binomial(n,k):
    if n<=k:
        return("n debe ser Mayor que k")
    elif k==0:
        return 1
    else:
        return int(((factorr(n))/((factorr(k))*(factorr(n-k)))))
        # n Debe ser mayor que k siempre!
```

```python
binomial(7,6)
```

4. Números primos

Es posible escribir un programa bastante rapido para calcular números primos siguiendo las siguientes observaciones:

a) Un numero n es primo, si no tiene factores primos menores que n. Por lo tanto, solo necesitamos comprobar si es divisible por otros numeros primos.

```python
n = int(input("Numero a probar: "))

L=[]
p=[]
for i in range(1,n):
    if i%2!=0:
        p.append(i)

# print(p)

for x in range(len(p)):
    if n%p[x]==0:
        # print(p[x])
        L.append(p[x])

print(L)
if L==[1] and n%2!=0:
    print("Es Primo")
else:
    print("No es Primo")
        # Si no incluyo el n%2!=0, el codigo me regresa que algunos numeros pares como el 4 u 8 tambien son primos!
```

a) Anteriormente, en clase encontramos los numeros de Catalán ́Cn. Si hacemos una pequena modificación, la definición de Cn se puede reescribir de la siguiente forma:

```python
def Cat(n):
    if n==0:
        return 1
    elif n<0:
        return print("n debe ser positiva")
    elif n>0:
        return (((4*n)-2)/(n+1))*Cat(n-1)
```
