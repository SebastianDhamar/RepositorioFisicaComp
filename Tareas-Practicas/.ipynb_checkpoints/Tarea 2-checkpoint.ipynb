a) Utiliza esta formula para escribir una funcion llamada  ́ binomial(n,k) (o como tu
quieras) que calcule el coeficiente binomial para un n y k dados. Asegurate de que  ́
tu funcion devuelva la respuesta en forma de un n  ́ umero entero (no flotante) y pro-  ́
porcione el valor correcto de 1 para el caso en que k = 0.

```python
def factorr(n):
    if n == 1:
        return 1
    elif n==0:
        return 1
    else:
        return n*factorr(n-1)
```

```python
def binomial(n,k):
    if k==0:
        return 1
    else:
        return (factorr(n)/(factorr(k)*factorr(n-k)))
```

```python
binomial(5,6)
```
